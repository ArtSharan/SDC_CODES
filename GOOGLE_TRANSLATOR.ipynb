```python
!pip install googletrans==4.0.0-rc1
# Importing the Translator from googletrans
from googletrans import Translator

# Create a Translator object
translator = Translator()

# Function to translate text
def translate_text(text, target_lang='en'):
    """
    Translates the input text to the target language.

    Parameters:
    text (str): The text to be translated.
    target_lang (str): The target language code (e.g., 'en' for English, 'fr' for French).

    Returns:
    str: The translated text.
    """
    translated = translator.translate(text, dest=target_lang)
    return translated.text

# Function to detect the language of the input text
def detect_language(text):
    """
    Detects the language of the input text.

    Parameters:
    text (str): The text whose language needs to be detected.

    Returns:
    str: The detected language code (e.g., 'en' for English).
    """
    detected = translator.detect(text)
    return detected.lang

# Example usage:
# Text to translate
text_to_translate = "Bonjour tout le monde"

# Detect the language of the input text
detected_lang = detect_language(text_to_translate)
print(f"Detected language: {detected_lang}")

# Translate the text to English
translated_text = translate_text(text_to_translate, target_lang='en')
print(f"Original text: {text_to_translate}")
print(f"Translated text: {translated_text}")

# Translate the text to French (from English)
translated_to_french = translate_text(translated_text, target_lang='fr')
print(f"Translated back to French: {translated_to_french}")
```
