```python
!pip install transformers
!pip install torch
from transformers import pipeline

# Load the summarization pipeline from Hugging Face
summarizer = pipeline("summarization")

# Function to summarize text
def summarize_text(text):
    summary = summarizer(text, max_length=150, min_length=50, do_sample=False)
    return summary[0]['summary_text']

# Example usage:
text = """
In the world of artificial intelligence, Natural Language Processing (NLP) plays a critical role. It is a branch of AI that focuses on the interaction between computers and human languages. NLP encompasses various tasks such as machine translation, text summarization, sentiment analysis, and more. Over the years, NLP has evolved from simple rule-based methods to complex machine learning models that can understand and generate human language. With the advent of deep learning, particularly transformer-based models like BERT, GPT, and T5, NLP has achieved remarkable progress. These models are trained on massive datasets and are capable of understanding context, nuances, and even generating creative text. The applications of NLP are vast, ranging from chatbots and virtual assistants to content generation and customer service automation.
"""

# Summarize the text
summary = summarize_text(text)
print(summary)
```
