```python
!pip install -q langchain openai gradio
import openai
from langchain.prompts import PromptTemplate
from langchain.llms import OpenAI
from langchain.chains import LLMChain
import gradio as gr

# ====== Set OpenAI API Key ======
openai.api_key = "your-openai-api-key"  # Replace with your OpenAI API key

# ====== Create a Medical Diagnosis Prompt Template ======
prompt_template = PromptTemplate(
    input_variables=["symptoms"],
    template="""
    You are an AI-powered medical assistant. Given the following symptoms, provide a list of possible diagnoses along with potential causes and relevant insights.

    Symptoms: {symptoms}

    Please also mention any red flags or urgent medical conditions that might be inferred from the symptoms.
    """
)

# ====== Initialize OpenAI LLM ======
llm = OpenAI(temperature=0.5)

# ====== Create Medical Diagnosis Chain ======
diagnosis_chain = LLMChain(llm=llm, prompt=prompt_template)

# ====== Diagnosis Function ======
def diagnose_symptoms(symptoms):
    diagnosis = diagnosis_chain.run({"symptoms": symptoms})
    return diagnosis

# ====== Gradio Interface for User Interaction ======
def analyze_symptoms(symptoms_input):
    diagnosis_result = diagnose_symptoms(symptoms_input)
    return diagnosis_result

# Create a simple Gradio interface
gr.Interface(
    fn=analyze_symptoms,
    inputs=gr.Textbox(lines=5, placeholder="Enter the symptoms (e.g., headache, fever, nausea)"),
    outputs="text",
    title="🩺 AI-Powered Medical Diagnosis Assistant",
    description="Enter a list of symptoms, and get a possible diagnosis along with relevant insights and potential causes."
).launch()
```
