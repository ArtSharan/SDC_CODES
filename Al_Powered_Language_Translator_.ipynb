```python
!pip install -q langchain openai gradio
import openai
from langchain.prompts import PromptTemplate
from langchain.llms import OpenAI
from langchain.chains import LLMChain
import gradio as gr

# ====== Set OpenAI API Key ======
openai.api_key = "your-openai-api-key"  # Replace with your OpenAI API key

# ====== Create Translation Prompt Template ======
prompt_template = PromptTemplate(
    input_variables=["text", "source_lang", "target_lang"],
    template="""
    Translate the following text from {source_lang} to {target_lang}:
    {text}
    """
)

# ====== Initialize OpenAI LLM ======
llm = OpenAI(temperature=0.3)

# ====== Create Translation Chain ======
translation_chain = LLMChain(llm=llm, prompt=prompt_template)

# ====== Translation Function ======
def translate_text(text, source_lang, target_lang):
    translation = translation_chain.run({"text": text, "source_lang": source_lang, "target_lang": target_lang})
    return translation

# ====== Gradio Interface for Translation ======
gr.Interface(
    fn=translate_text,
    inputs=[
        gr.Textbox(lines=5, placeholder="Enter text to translate..."),
        gr.Textbox(lines=1, placeholder="Source language (e.g., English)"),
        gr.Textbox(lines=1, placeholder="Target language (e.g., Spanish)")
    ],
    outputs="text",
    title="🌍 AI-Powered Multilingual Translator",
    description="Translate text from one language to another in real-time using OpenAI's GPT model."
).launch()
```
