```python
!pip install -q langchain openai transformers torch huggingface_hub
import os
from langchain.llms import OpenAI
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline

# ====== Set Your Keys ======
os.environ["OPENAI_API_KEY"] = "your-openai-key"
os.environ["HUGGINGFACEHUB_API_TOKEN"] = "your-huggingface-token"

# ====== Prompt Template ======
prompt_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms."
)

# ====== LLM 1: OpenAI via LangChain ======
llm_openai = OpenAI(temperature=0.7, model_name="gpt-3.5-turbo-instruct")
chain_openai = LLMChain(llm=llm_openai, prompt=prompt_template)

# ====== LLM 2: HuggingFace Transformers (Direct, No LangChain) ======
model_name = "google/flan-t5-large"
hf_pipe = pipeline("text2text-generation", model=model_name)

# ====== Topic to Test ======
topic = "quantum computing"
formatted_prompt = f"Explain {topic} in simple terms."

# ====== Run OpenAI via LangChain ======
output_openai = chain_openai.run(topic)

# ====== Run Hugging Face Model Directly ======
output_hf = hf_pipe(formatted_prompt, max_length=100, temperature=0.7)[0]["generated_text"]

# ====== Results ======
print("🧠 OpenAI via LangChain:")
print(output_openai)

print("\n🤖 Hugging Face Direct:")
print(output_hf)
```
