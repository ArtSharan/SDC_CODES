```python
!pip install -q transformers gradio
from transformers import pipeline
import gradio as gr

# Initialize Hugging Face summarization pipeline (BART or T5 model)
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# Function to summarize the article using Hugging Face model
def summarize_article_hf(article_text):
    summary = summarizer(article_text, max_length=150, min_length=50, do_sample=False)
    return summary[0]['summary_text']

# Create Gradio interface for user interaction
gr.Interface(fn=summarize_article_hf,
             inputs=gr.Textbox(lines=10, placeholder="Enter your news article here..."),
             outputs="text",
             title="📰 News Summarizer",
             description="Enter a news article and get a concise summary.") \
    .launch()
```
